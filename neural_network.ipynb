```python
#imports
import numpy as np
```

```python
# Synthetic regression data
rng = np.random.default_rng(42)

n_samples = 600

X = rng.uniform(-2, 2, size=(n_samples, 2))
noise = rng.normal(0, 0.20, size=n_samples)

y = (
    1.5 * X[:, 0] ** 2
    + 0.8 * X[:, 1]
    - 0.4 * X[:, 0] * X[:, 1]
    + noise
).reshape(-1, 1)
```

```python
def relu(z):
    """Apply the ReLU activation function elementwise."""
    return np.maximum(0, z)
```

```python
# Network architecture
input_size = 2
hidden_size = 8
output_size = 1

W1 = rng.uniform(-1, 1, size=(input_size, hidden_size))
b1 = np.zeros((1, hidden_size))

W2 = rng.uniform(-1, 1, size=(hidden_size, output_size))
b2 = np.zeros((1, output_size))


def forward(X, W1, b1, W2, b2):
    """
    Perform a forward pass through a one-hidden-layer neural network.

    Parameters
    ----------
    X : np.ndarray
        Input data with shape (n_samples, input_size).
    W1 : np.ndarray
        Weights of the hidden layer.
    b1 : np.ndarray
        Biases of the hidden layer.
    W2 : np.ndarray
        Weights of the output layer.
    b2 : np.ndarray
        Bias of the output layer.

    Returns
    -------
    Z1 : np.ndarray
        Pre-activation values of the hidden layer.
    A1 : np.ndarray
        Hidden-layer activations after ReLU.
    y_hat : np.ndarray
        Model predictions.
    """
    Z1 = X @ W1 + b1
    A1 = relu(Z1)
    y_hat = A1 @ W2 + b2

    return Z1, A1, y_hat
```

```python

Z1, A1, y_hat = forward(X, W1, b1, W2, b2)
assert Z1.shape == (600, 8)
assert A1.shape == (600, 8)
assert y_hat.shape == (600, 1)

assert np.all(A1 >= 0)

sample_X = X[0:4]
n = len(sample_X)
Z1, A1, y_hat = forward(sample_X,W1,b1,W2,b2)
assert y_hat.shape == (n,1)

print("Forward pass tests passed.")
```
